# Mini-GNN Segment Embedder — Cosine Similarity Analysis

Inspect the trained `SegmentGNN` by computing cosine similarity distributions for
**true segment pairs** (same particle) vs **negative pairs** (different particles).

**Part 1** — GNN-inferred CC segments (exactly as used at inference time)  
**Part 2** — Ground truth segments (`hit_segment_id`)  
**Part 3** — GT physics arcs (ingoing/outgoing split at radial turnaround)

```python
import sys
from pathlib import Path

import numpy as np
import torch
import matplotlib.pyplot as plt
from torch_geometric.data import Batch, Data
from tqdm import tqdm

SCRIPT_DIR    = Path().resolve()
PIPELINE_ROOT = SCRIPT_DIR.parent
WORKSPACE_ROOT = PIPELINE_ROOT.parent

sys.path.insert(0, str(WORKSPACE_ROOT / 'acorn'))
sys.path.insert(0, str(PIPELINE_ROOT))


from low_pt_custom_utils.track_evaluation_utils import PDG_TO_SPECIES
from low_pt_custom_utils.track_evaluation_utils import SPECIES_COLORS, SPECIES_MARKERS

from low_pt_custom_utils.mini_gnn_segment_embedding import (
    load_segment_gnn,
    segment_to_pyg,
    get_segment_particle_id,
)
from low_pt_custom_utils.segment_matching import (
    extract_segments_from_cc,
    extract_segments_from_ground_truth,
)
from low_pt_custom_utils.wrangler_utils import extract_segments_with_wrangler
```

```python
MODEL_PATH = (
    PIPELINE_ROOT
    / 'data_100_mixed/track_building/mini_GNN_checkpoints/mini_gnn_segment_embedder_val_loss=1.2621.pt'
)

device = 'cuda' if torch.cuda.is_available() else 'cpu'
model  = load_segment_gnn(str(MODEL_PATH), device=device)

ckpt = torch.load(str(MODEL_PATH), map_location='cpu', weights_only=True)
print(f'Model: {MODEL_PATH.name}')
print(f'  hidden_dim={ckpt["hidden_dim"]}, emb_dim={ckpt["emb_dim"]}, '
      f'n_layers={ckpt["n_layers"]}, dropout={ckpt["dropout"]}')
print(f'  node_scales={ckpt["node_scales"]}')
print(f'  device={device}')
```

```python
# --- Configuration (matches gnn_segment_matching.yaml) ---
SCORE_CUT         = 0.4
USE_WRANGLER      = True   # compare CC vs wrangler
OUTER_R_THRESHOLD = 1000.0   # mm — outermost hit >= this → complete track, excluded from matching

DATA_DIR    = PIPELINE_ROOT / 'data' / 'gnn_stage' / 'valset'
event_paths = sorted(DATA_DIR.glob('*.pyg'))
print(f'Events:            {len(event_paths)}')
print(f'Score cut:         {SCORE_CUT}')
print(f'outer_r threshold: {OUTER_R_THRESHOLD} mm')
```

```python
# --- Shared helpers ---

def embed_segments(segments, graph, model, device):
    """Return (embeddings, particle_ids) for a list of SegmentInfo."""
    node_scales = tuple(model.node_scales)
    seg_graphs  = [segment_to_pyg(seg, graph, node_scales=node_scales) for seg in segments]
    batch_data  = Batch.from_data_list(seg_graphs).to(device)
    with torch.no_grad():
        embs = model(batch_data.x, batch_data.edge_index, batch_data.batch)
    embs = embs.cpu()
    pids = np.array([get_segment_particle_id(seg, graph) for seg in segments])
    return embs, pids


def true_pair_cosine_sims(embs, pids):
    """Cosine similarities for all same-particle segment pairs (pid > 0)."""
    sim_matrix = (embs @ embs.T).numpy()
    sims = []
    n = len(pids)
    for i in range(n):
        if pids[i] == 0:
            continue
        for j in range(i + 1, n):
            if pids[j] == pids[i]:
                sims.append(sim_matrix[i, j])
    return np.array(sims)


def neg_pair_cosine_sims(embs, pids):
    """Cosine similarities for all different-particle segment pairs."""
    sim_matrix = (embs @ embs.T).numpy()
    sims = []
    n = len(pids)
    for i in range(n):
        if pids[i] == 0:
            continue
        for j in range(i + 1, n):
            if pids[j] != pids[i] and pids[j] != 0:
                sims.append(sim_matrix[i, j])
    return np.array(sims)


bins = np.linspace(-1, 1, 80)
print('Helpers defined')

def get_segments(graph):
    """Extract segments using CC + Wrangler (or plain CC if USE_WRANGLER=False)."""
    if USE_WRANGLER:
        segs, _ = extract_segments_with_wrangler(graph, SCORE_CUT)
        return segs
    return extract_segments_from_cc(graph, SCORE_CUT)
```

## Part 1 — GNN-inferred CC segments

Segments from Connected Components on GNN edge scores (`score > SCORE_CUT`).
Complete-track segments (`outer_r >= OUTER_R_THRESHOLD`) are excluded — identical to inference.

```python
# --- Part 1: all events, CC segments ---

cc_true_sims = []
cc_neg_sims  = []

for path in tqdm(event_paths, desc='CC segments'):
    graph = torch.load(path, map_location='cpu', weights_only=False)

    segments = get_segments(graph)
    segments = [s for s in segments if s.outer_r < OUTER_R_THRESHOLD]
    if len(segments) < 2:
        continue

    embs, pids = embed_segments(segments, graph, model, device)
    cc_true_sims.extend(true_pair_cosine_sims(embs, pids))
    cc_neg_sims.extend(neg_pair_cosine_sims(embs, pids))

cc_true_sims = np.array(cc_true_sims)
cc_neg_sims  = np.array(cc_neg_sims)

print(f'True pairs:     {len(cc_true_sims):>7,}')
print(f'Negative pairs: {len(cc_neg_sims):>7,}')
if len(cc_true_sims):
    print(f'\nTrue   mean={cc_true_sims.mean():.4f}  median={np.median(cc_true_sims):.4f}  std={cc_true_sims.std():.4f}')
if len(cc_neg_sims):
    print(f'Neg    mean={cc_neg_sims.mean():.4f}  median={np.median(cc_neg_sims):.4f}  std={cc_neg_sims.std():.4f}')
```

```python
fig, ax = plt.subplots(figsize=(7, 5))
if len(cc_neg_sims):
    ax.hist(cc_neg_sims,  bins=bins, alpha=0.45, color='steelblue',
            label=f'Negative pairs  (n={len(cc_neg_sims):,})', density=True)
if len(cc_true_sims):
    ax.hist(cc_true_sims, bins=bins, alpha=0.75, color='tomato',
            label=f'True pairs  (n={len(cc_true_sims):,})', density=True)

ax.set_xlabel('Cosine similarity', fontsize=18)
ax.set_ylabel('Probability density', fontsize=18)
ax.tick_params(labelsize=14)
ax.legend(fontsize=14)
ax.grid(True, alpha=0.3)
ax.set_xlim(-1, 1)
fig.tight_layout()

out_path = PIPELINE_ROOT / 'data' / 'visuals' / 'cosine_sim_cc_segments_global.pdf'
out_path.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(out_path, format='pdf', bbox_inches='tight', dpi=150)
print(f'Saved: {out_path}')

plt.show()
```

## Part 1.5 — Centroid Arc Check (Geometric Cut)

A purely geometric veto for false-positive segment matches.

**Idea:** for a looping particle, the outgoing and incoming arcs lie on the *same circle*.
Their hit centroids therefore sit on **opposite sides** of the circle center.
Two outgoing (or two incoming) arcs from *different* particles that happen to share
similar helix parameters will have centroids on the **same side**.

For a pair of segments A, B with fitted circle centers $(x_c^A, y_c^A)$ and $(x_c^B, y_c^B)$
and hit centroids $\bar{x}^A, \bar{x}^B$, define the centroid-arc vectors:

$$\mathbf{v}_A = \bar{\mathbf{x}}^A - \mathbf{c}^A, \qquad \mathbf{v}_B = \bar{\mathbf{x}}^B - \mathbf{c}^B$$

**Veto:** if both segments share a circle ($\mathbf{c}^A \approx \mathbf{c}^B$), use the shared center:

$$\text{reject if } \mathbf{v}_A \cdot \mathbf{v}_B > 0 \quad \text{(same arc side)}$$


```python
from collections import Counter
from low_pt_custom_utils.helix_fitting import fit_helix_to_segment

# --- Geometry check config ---
MAX_CENTER_DIST_MM = 200.0   # Require fitted circle centers to agree
MIN_HITS_FOR_FIT   = 3

# --- Fit helices and cache per-segment geometry once ---
hx = graph_single.hit_x.numpy()
hy = graph_single.hit_y.numpy()
hz = graph_single.hit_z.numpy()

def segment_geom(seg):
    """Return cached geometry for a segment, or None if fit is unusable."""
    if len(seg.hits) < MIN_HITS_FOR_FIT:
        return None
    sx = hx[seg.hits]
    sy = hy[seg.hits]
    sz = hz[seg.hits]

    h = fit_helix_to_segment(sx, sy, sz)
    if h.fit_quality in ("none", "poor"):
        return None

    centroid = np.array([sx.mean(), sy.mean()], dtype=float)
    center = np.array([h.xc, h.yc], dtype=float)
    return {"centroid": centroid, "center": center, "pid": int(get_segment_particle_id(seg, graph_single))}

seg_geom = [segment_geom(s) for s in segs_cc]

def arc_dot(i, j, max_center_dist=MAX_CENTER_DIST_MM):
    """
    Dot product of centroid vectors relative to a shared center.
    Returns NaN if fit is missing or circles are not geometrically compatible.
    """
    gi = seg_geom[i]
    gj = seg_geom[j]
    if gi is None or gj is None:
        return np.nan

    center_dist = np.linalg.norm(gi["center"] - gj["center"])
    if center_dist > max_center_dist:
        return np.nan

    shared_center = 0.5 * (gi["center"] + gj["center"])
    vi = gi["centroid"] - shared_center
    vj = gj["centroid"] - shared_center

    vi /= np.linalg.norm(vi) + 1e-9
    vj /= np.linalg.norm(vj) + 1e-9
    return float(np.dot(vi, vj))

# --- Pair loop with explicit accounting ---
n = len(segs_cc)
true_dots = []
neg_dots = []
n_pairs_total = 0
n_pairs_center_rejected = 0
n_pairs_fit_rejected = 0

for i in range(n):
    pid_i = int(pids_single[i])
    if pid_i == 0:
        continue
    for j in range(i + 1, n):
        pid_j = int(pids_single[j])
        if pid_j == 0:
            continue

        n_pairs_total += 1

        gi = seg_geom[i]
        gj = seg_geom[j]
        if gi is None or gj is None:
            n_pairs_fit_rejected += 1
            continue

        cd = np.linalg.norm(gi["center"] - gj["center"])
        if cd > MAX_CENTER_DIST_MM:
            n_pairs_center_rejected += 1
            continue

        d = arc_dot(i, j)
        if np.isnan(d):
            continue

        if pid_j == pid_i:
            true_dots.append(d)
        else:
            neg_dots.append(d)

true_dots = np.array(true_dots)
neg_dots = np.array(neg_dots)

print(f"Segments in event: {n}")
print(f"Total non-noise segment pairs checked: {n_pairs_total}")
print(f"Rejected (fit missing/poor): {n_pairs_fit_rejected}")
print(f"Rejected (center mismatch > {MAX_CENTER_DIST_MM:.1f} mm): {n_pairs_center_rejected}")
print(f"Accepted true pairs: {len(true_dots)}")
print(f"Accepted neg  pairs: {len(neg_dots)}")

if len(true_dots):
    print(f"\nTrue: dot<0 = {(true_dots < 0).mean():.1%}, dot>0 = {(true_dots > 0).mean():.1%}")
if len(neg_dots):
    print(f"Neg : dot<0 = {(neg_dots < 0).mean():.1%}, dot>0 = {(neg_dots > 0).mean():.1%}")

# --- Key diagnostic: particles with >2 segments create many same-side true combinations ---
pid_counts = Counter(int(pid) for pid in pids_single if int(pid) > 0)
multi_pid_counts = [c for c in pid_counts.values() if c > 1]
print("\nParticles by number of segments (pid>0):")
for k in sorted(Counter(pid_counts.values())):
    print(f"  {k} segment(s): {Counter(pid_counts.values())[k]} particles")

if multi_pid_counts:
    print(f"Particles with >1 segment: {sum(c > 1 for c in pid_counts.values())}")

# For each multi-segment particle, report all-pairs opposite rate and the most-opposite pair
print("\nPer-particle true-pair diagnostics (multi-segment only):")
print(f"  {'particle_id':>15}  {'n_seg':>5}  {'n_pairs':>7}  {'frac(dot<0)':>12}  {'min_dot':>8}")
print('  ' + '-' * 62)
for pid, c in sorted(pid_counts.items()):
    if c < 2:
        continue
    idx = np.where(pids_single == pid)[0]
    dots = []
    for a in range(len(idx)):
        for b in range(a + 1, len(idx)):
            d = arc_dot(int(idx[a]), int(idx[b]))
            if not np.isnan(d):
                dots.append(d)
    if not dots:
        continue
    dots = np.array(dots)
    print(f"  {pid:>15}  {c:>5}  {len(dots):>7}  {(dots < 0).mean():>12.1%}  {dots.min():>8.3f}")

# --- Plot ---
fig, ax = plt.subplots(figsize=(8, 5))
dot_bins = np.linspace(-1, 1, 60)
if len(neg_dots):
    ax.hist(
        neg_dots, bins=dot_bins, alpha=0.45, color="steelblue", density=True,
        label=f"Negative pairs  (n={len(neg_dots)})"
    )
if len(true_dots):
    ax.hist(
        true_dots, bins=dot_bins, alpha=0.75, color="tomato", density=True,
        label=f"True pairs  (n={len(true_dots)})"
    )
ax.axvline(0, color="black", linestyle="--", lw=1.5, label="veto threshold (dot = 0)")
ax.set_xlabel("Centroid-arc dot product  ($\\mathbf{v}_A \\cdot \\mathbf{v}_B$)", fontsize=13)
ax.set_ylabel("Density", fontsize=13)
ax.set_title(f"Geometric arc check — event {event_id}", fontsize=13)
ax.legend(fontsize=11)
ax.set_xlim(-1, 1)
fig.tight_layout()
plt.show()
```

### Feature 2 — Arc Orientation (Chord Side)

For each segment draw the chord from **first to last time-ordered hit**. Use the signed cross product to determine which side of the chord the remaining hits lie on (+1 = left, −1 = right).

**Hypothesis:**
- **True pair** (outgoing + return arc of the same circle): the return chord is roughly anti-parallel to the outgoing chord, so the arc bulges on **opposite sides** → orientation signs differ.
- **Confuser pair** (two outgoing arcs from different particles): both chords point the same way, both arcs bulge the **same side** → orientation signs match → veto.

This is essentially a discrete version of the chord angle check below.


```python

# --- Part 1.5b: Arc Orientation (Chord Side) ---
# For each segment, draw the chord from the first to the last time-ordered hit.
# Compute the signed cross product (chord × (hit - first)).z for each intermediate hit.
# The mean sign tells us which side of the chord the arc bulges toward.
#
# True pair (outgoing + return): the particle traverses the same circle in both arcs,
#   but the chord for the return arc is oriented roughly opposite to the outgoing chord.
#   → The arc bulges to OPPOSITE sides → different orientation signs.
# Confuser pair (both outgoing, different particles): both chords point the same way,
#   both arcs bulge the same way → SAME orientation sign.
#
# Prediction: veto pairs with equal orientation sign.

def arc_orientation_sign(seg, hx, hy):
    """
    +1 or -1: which side of the first→last chord do the intermediate hits lie?
    Returns None if < 3 hits (no intermediate points).
    """
    hits = seg.hits
    if len(hits) < 3:
        return None
    p0 = np.array([hx[hits[0]],  hy[hits[0]]])
    p1 = np.array([hx[hits[-1]], hy[hits[-1]]])
    chord = p1 - p0
    chord_len = np.linalg.norm(chord)
    if chord_len < 1e-3:
        return None

    crosses = []
    for h in hits[1:-1]:
        ph = np.array([hx[h], hy[h]])
        v = ph - p0
        cross_z = chord[0] * v[1] - chord[1] * v[0]
        crosses.append(cross_z)

    mean_cross = np.mean(crosses)
    if abs(mean_cross) < 1e-6:
        return None
    return int(np.sign(mean_cross))

seg_orientation = [arc_orientation_sign(s, hx, hy) for s in segs_cc]

true_same = 0; true_diff = 0
neg_same  = 0; neg_diff  = 0
true_skip = 0; neg_skip  = 0

for i in range(n):
    pid_i = int(pids_single[i])
    if pid_i == 0:
        continue
    oi = seg_orientation[i]
    for j in range(i + 1, n):
        pid_j = int(pids_single[j])
        if pid_j == 0:
            continue
        oj = seg_orientation[j]

        if oi is None or oj is None:
            if pid_i == pid_j: true_skip += 1
            else:              neg_skip  += 1
            continue

        same = (oi == oj)
        if pid_i == pid_j:
            if same: true_same += 1
            else:    true_diff += 1
        else:
            if same: neg_same += 1
            else:    neg_diff += 1

n_true = true_same + true_diff
n_neg  = neg_same  + neg_diff

print(f"Arc orientation (chord-side) check  [skipped: {true_skip} true, {neg_skip} neg — too few hits]")
print(f"True pairs:  same-side = {true_same}/{n_true} ({true_same/max(n_true,1):.1%}),  diff-side = {true_diff}/{n_true} ({true_diff/max(n_true,1):.1%})")
print(f"Neg  pairs:  same-side = {neg_same}/{n_neg}  ({neg_same/max(n_neg,1):.1%}),  diff-side = {neg_diff}/{n_neg}  ({neg_diff/max(n_neg,1):.1%})")
print()
print(f"Veto 'same-side' pairs:")
print(f"  Removes {neg_same/max(n_neg,1):.1%} of neg pairs  ✓")
print(f"  Wrongly removes {true_same/max(n_true,1):.1%} of true pairs  ✗")

# Bar chart summary
labels = ['True pairs', 'Negative pairs']
same_fracs = [true_same / max(n_true, 1), neg_same / max(n_neg, 1)]
diff_fracs = [true_diff / max(n_true, 1), neg_diff / max(n_neg, 1)]

x = np.arange(len(labels))
width = 0.35
fig, ax = plt.subplots(figsize=(7, 4.5))
ax.bar(x - width/2, same_fracs, width, label='Same side (→ veto)',  color=['tomato',    'steelblue'], alpha=0.85)
ax.bar(x + width/2, diff_fracs, width, label='Diff side (→ allow)', color=['lightcoral', 'lightblue'],  alpha=0.85)
ax.set_xticks(x); ax.set_xticklabels(labels, fontsize=12)
ax.set_ylabel('Fraction of pairs', fontsize=12)
ax.set_title(f'Arc orientation — event {event_id}', fontsize=13)
ax.set_ylim(0, 1)
ax.legend(fontsize=10)
fig.tight_layout()
plt.show()
```

### Feature 3 — Chord Angle Difference

For each segment draw the chord from its **first to last time-ordered hit**.

- **True pair** (outgoing + return of the same particle): the particle physically reverses direction at the turnaround point, so the two chords point in **opposite** directions → angle difference near ±π.
- **Confuser pair** (two different particles' outgoing segments): both chords point in the **same** direction → angle difference near 0.

The shaded green bands mark the anti-parallel region (|Δθ| > π/2).


```python

# --- Part 1.5c: Chord Angle Difference ---
# For each segment, the chord is the vector from first to last time-ordered hit.
# True pair (outgoing + return): the particle reverses direction, so the chords
#   point in roughly OPPOSITE directions → angle difference near ±π.
# Confuser pair (both outgoing, different particles): chords point in the SAME
#   direction → angle difference near 0.

def chord_angle(seg, hx, hy):
    """Angle of the first→last chord in the x-y plane, in radians."""
    hits = seg.hits
    if len(hits) < 2:
        return None
    dx = hx[hits[-1]] - hx[hits[0]]
    dy = hy[hits[-1]] - hy[hits[0]]
    chord_len = np.hypot(dx, dy)
    if chord_len < 1e-3:
        return None
    return float(np.arctan2(dy, dx))

seg_chord_angles = [chord_angle(s, hx, hy) for s in segs_cc]

true_angle_diffs = []
neg_angle_diffs  = []

for i in range(n):
    pid_i = int(pids_single[i])
    if pid_i == 0 or seg_chord_angles[i] is None:
        continue
    for j in range(i + 1, n):
        pid_j = int(pids_single[j])
        if pid_j == 0 or seg_chord_angles[j] is None:
            continue

        diff = ((seg_chord_angles[j] - seg_chord_angles[i]) + np.pi) % (2 * np.pi) - np.pi

        if pid_i == pid_j:
            true_angle_diffs.append(diff)
        else:
            neg_angle_diffs.append(diff)

true_angle_diffs = np.array(true_angle_diffs)
neg_angle_diffs  = np.array(neg_angle_diffs)

# Summary: fraction near anti-parallel (|diff| > π/2)
thresh = np.pi / 2
print(f"Chord angle difference")
print(f"True pairs: |diff| > π/2 = {(np.abs(true_angle_diffs) > thresh).mean():.1%}  (expected: high — opposite directions)")
print(f"Neg  pairs: |diff| > π/2 = {(np.abs(neg_angle_diffs) > thresh).mean():.1%}  (expected: low  — same direction)")

angle_bins = np.linspace(-np.pi, np.pi, 40)
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(neg_angle_diffs,  bins=angle_bins, alpha=0.45, color='steelblue', density=True,
        label=f'Negative pairs  (n={len(neg_angle_diffs)})')
ax.hist(true_angle_diffs, bins=angle_bins, alpha=0.75, color='tomato',    density=True,
        label=f'True pairs  (n={len(true_angle_diffs)})')
ax.axvspan(-np.pi, -np.pi / 2, color='green', alpha=0.07, label='anti-parallel region')
ax.axvspan( np.pi / 2,  np.pi, color='green', alpha=0.07)
ax.set_xlabel('Chord angle difference (rad)', fontsize=13)
ax.set_ylabel('Density', fontsize=13)
ax.set_title(f'Chord angle difference — event {event_id}', fontsize=13)
ax.set_xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
ax.set_xticklabels(['-π', '-π/2', '0', 'π/2', 'π'])
ax.legend(fontsize=11)
fig.tight_layout()
plt.show()
```

### Feature 4 — Endpoint Proximity

A true pair (outgoing arc + return arc of the same particle) **share a turnaround point**: the last hit of the outgoing segment and the first hit of the return segment are physically close.

Two confuser segments from different particles have no such shared endpoint — their endpoints are uncorrelated.

Metric: `min(d(A_start,B_start), d(A_start,B_end), d(A_end,B_start), d(A_end,B_end))` in mm.


```python

# --- Part 1.5d: Endpoint Proximity ---
# For a true pair (outgoing + return of the same particle), the outgoing segment ends
# and the return segment begins near the turnaround point — they share an endpoint region.
# For confuser pairs (two different particles' outgoing segments), no such proximity.
#
# For each pair, compute the minimum of the 4 endpoint-endpoint distances:
#   min( d(A_start,B_start), d(A_start,B_end), d(A_end,B_start), d(A_end,B_end) )

def segment_endpoints(seg, hx, hy):
    """Return (first_xy, last_xy) of a time-ordered segment."""
    if len(seg.hits) < 2:
        return None
    p0 = np.array([hx[seg.hits[0]],  hy[seg.hits[0]]])
    p1 = np.array([hx[seg.hits[-1]], hy[seg.hits[-1]]])
    return p0, p1

seg_endpoints = [segment_endpoints(s, hx, hy) for s in segs_cc]

true_min_ep_dist = []
neg_min_ep_dist  = []

for i in range(n):
    pid_i = int(pids_single[i])
    if pid_i == 0 or seg_endpoints[i] is None:
        continue
    for j in range(i + 1, n):
        pid_j = int(pids_single[j])
        if pid_j == 0 or seg_endpoints[j] is None:
            continue

        ei = seg_endpoints[i]
        ej = seg_endpoints[j]
        dists = [
            np.linalg.norm(ei[0] - ej[0]),
            np.linalg.norm(ei[0] - ej[1]),
            np.linalg.norm(ei[1] - ej[0]),
            np.linalg.norm(ei[1] - ej[1]),
        ]
        min_d = min(dists)

        if pid_i == pid_j:
            true_min_ep_dist.append(min_d)
        else:
            neg_min_ep_dist.append(min_d)

true_min_ep_dist = np.array(true_min_ep_dist)
neg_min_ep_dist  = np.array(neg_min_ep_dist)

print(f"Endpoint proximity")
print(f"True pairs: median min-dist = {np.median(true_min_ep_dist):.1f} mm,  < 50 mm: {(true_min_ep_dist < 50).mean():.1%}")
print(f"Neg  pairs: median min-dist = {np.median(neg_min_ep_dist):.1f} mm,  < 50 mm: {(neg_min_ep_dist < 50).mean():.1%}")

ep_bins = np.linspace(0, 500, 60)
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(neg_min_ep_dist,  bins=ep_bins, alpha=0.45, color='steelblue', density=True,
        label=f'Negative pairs  (n={len(neg_min_ep_dist)})')
ax.hist(true_min_ep_dist, bins=ep_bins, alpha=0.75, color='tomato',    density=True,
        label=f'True pairs  (n={len(true_min_ep_dist)})')
ax.set_xlabel('Min endpoint-to-endpoint distance (mm)', fontsize=13)
ax.set_ylabel('Density', fontsize=13)
ax.set_title(f'Endpoint proximity — event {event_id}', fontsize=13)
ax.legend(fontsize=11)
fig.tight_layout()
plt.show()
```

### Feature 5 — Per-radius Position Difference

For a pair of segments, sample both at common radii and compute the 2D distance $\|(\Delta x, \Delta y)\|$ at each $r$.

- **True pair** (outgoing + return): same circle, different angular region → finite separation at every $r$.
- **Confuser pair** (two nearly identical outgoing segments): nearly co-located at every $r$ → difference ≈ 0 everywhere.

Sampling strategy: sort each segment's hits by $r$, then match by nearest-$r$ in the other segment. Hard negatives are filtered to pairs where the fitted circle centers agree (center dist < 60 mm) — these are the actual confusers the GNN struggles with.


```python

# --- Part 1.5e: Per-radius Position Difference (overlap only) ---
# Only compare hit positions where BOTH segments have a hit at a similar radius.
# r_tol: maximum |r_A - r_B| to consider two hits "at the same layer".

R_TOL_MM = 15.0   # detector layers are spaced further than this, so this picks same-layer hits

hr = graph_single.hit_r.numpy()

def seg_sorted_by_r(seg):
    idx = np.array(seg.hits)
    rs = hr[idx]; xs = hx[idx]; ys = hy[idx]
    order = np.argsort(rs)
    return rs[order], xs[order], ys[order]

def pos_diff_vs_r(seg_i, seg_j, r_tol=R_TOL_MM):
    """
    For each hit in seg_i, find the nearest-r hit in seg_j.
    Only include the pair if |r_A - r_B| < r_tol (both segments have a hit at that layer).
    """
    ri, xi, yi = seg_sorted_by_r(seg_i)
    rj, xj, yj = seg_sorted_by_r(seg_j)
    r_out, d_out = [], []
    for r, x, y in zip(ri, xi, yi):
        k = int(np.argmin(np.abs(rj - r)))
        if np.abs(rj[k] - r) < r_tol:
            d_out.append(np.hypot(x - xj[k], y - yj[k]))
            r_out.append(r)
    return np.array(r_out), np.array(d_out)

# --- Collect pairs: true and hard-negative (center_dist < MAX_CENTER_DIST_MM) ---
true_curves     = []
hard_neg_curves = []

for i in range(n):
    pid_i = int(pids_single[i])
    if pid_i == 0 or seg_geom[i] is None:
        continue
    for j in range(i + 1, n):
        pid_j = int(pids_single[j])
        if pid_j == 0 or seg_geom[j] is None:
            continue

        center_dist = np.linalg.norm(seg_geom[i]["center"] - seg_geom[j]["center"])
        r_arr, d_arr = pos_diff_vs_r(segs_cc[i], segs_cc[j])
        if len(d_arr) == 0:
            continue   # no overlapping layers

        if pid_i == pid_j:
            true_curves.append((r_arr, d_arr))
        elif center_dist < MAX_CENTER_DIST_MM:
            hard_neg_curves.append((r_arr, d_arr))

print(f"True pairs with overlapping layers:      {len(true_curves)}")
print(f"Hard-negative pairs (center_dist < {MAX_CENTER_DIST_MM:.0f} mm): {len(hard_neg_curves)}")
print(f"r_tol = {R_TOL_MM} mm")

# --- Plot individual curves ---
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, curves, color, label in [
    (axes[0], true_curves,     'tomato',    'True pairs (outgoing + return)'),
    (axes[1], hard_neg_curves, 'steelblue', f'Hard negatives (center dist < {MAX_CENTER_DIST_MM:.0f} mm)'),
]:
    for r_arr, d_arr in curves:
        ax.plot(r_arr, d_arr, color=color, alpha=0.4, lw=1.2)
    ax.set_xlabel('Hit radius $r$ (mm)', fontsize=12)
    ax.set_ylabel('2D position diff $\\|\\Delta(x,y)\\|$ (mm)', fontsize=12)
    ax.set_title(label, fontsize=12)
    ax.set_xlim(0, None); ax.set_ylim(0, None)
fig.suptitle(f'Per-radius position difference (overlap only) — event {event_id}', fontsize=13)
fig.tight_layout(); plt.show()

# --- Summary histogram: mean diff per pair ---
true_mean_diffs     = [d.mean() for _, d in true_curves]
hard_neg_mean_diffs = [d.mean() for _, d in hard_neg_curves]

fig2, ax2 = plt.subplots(figsize=(8, 4.5))
bins2 = np.linspace(0, 400, 60)
if hard_neg_mean_diffs:
    ax2.hist(hard_neg_mean_diffs, bins=bins2, alpha=0.55, color='steelblue', density=True,
             label=f'Hard negatives (n={len(hard_neg_mean_diffs)})')
if true_mean_diffs:
    ax2.hist(true_mean_diffs, bins=bins2, alpha=0.75, color='tomato', density=True,
             label=f'True pairs (n={len(true_mean_diffs)})')
ax2.set_xlabel('Mean 2D position diff at overlapping layers (mm)', fontsize=12)
ax2.set_ylabel('Density', fontsize=12)
ax2.set_title(f'Mean per-r difference (overlap only) — event {event_id}', fontsize=12)
ax2.legend(fontsize=11)
fig2.tight_layout(); plt.show()
```

```python

# --- Valset: Per-radius Position Difference (vectorised, no fitting, all negatives) ---

def mean_pos_diff(ri, xi, yi, rj, xj, yj, r_tol=R_TOL_MM):
    """
    Vectorised: build (Hi x Hj) r-distance matrix, pick nearest-r match per hit in i,
    keep only matches within r_tol, return mean 2D distance. Returns nan if no overlap.
    """
    dr = np.abs(ri[:, None] - rj[None, :])   # (Hi, Hj)
    nearest = np.argmin(dr, axis=1)           # (Hi,)
    mask = dr[np.arange(len(ri)), nearest] < r_tol
    if not mask.any():
        return np.nan
    dx = xi[mask] - xj[nearest[mask]]
    dy = yi[mask] - yj[nearest[mask]]
    return float(np.mean(np.hypot(dx, dy)))

val_true_means = []
val_neg_means  = []

for path in tqdm(event_paths, desc='Valset pos-diff check'):
    graph = torch.load(path, map_location='cpu', weights_only=False)
    segs  = get_segments(graph)
    segs  = [s for s in segs if s.outer_r < OUTER_R_THRESHOLD]
    if len(segs) < 2:
        continue

    _hr = graph.hit_r.numpy()
    _hx = graph.hit_x.numpy()
    _hy = graph.hit_y.numpy()
    pids = np.array([get_segment_particle_id(s, graph) for s in segs])

    # Pre-sort each segment's hits by r once
    sorted_segs = []
    for s in segs:
        idx = np.array(s.hits)
        order = np.argsort(_hr[idx])
        sorted_segs.append((_hr[idx][order], _hx[idx][order], _hy[idx][order]))

    n_s = len(segs)
    for i in range(n_s):
        pid_i = int(pids[i])
        if pid_i == 0:
            continue
        ri, xi, yi = sorted_segs[i]
        for j in range(i + 1, n_s):
            pid_j = int(pids[j])
            if pid_j == 0:
                continue
            rj, xj, yj = sorted_segs[j]

            mean_d = mean_pos_diff(ri, xi, yi, rj, xj, yj)
            if np.isnan(mean_d):
                continue

            if pid_i == pid_j:
                val_true_means.append(mean_d)
            else:
                val_neg_means.append(mean_d)

val_true_means = np.array(val_true_means)
val_neg_means  = np.array(val_neg_means)

print(f"Valset — true pairs:     {len(val_true_means):,}")
print(f"Valset — negative pairs: {len(val_neg_means):,}")
print()
for cut in [30, 50, 100, 150, 200]:
    tp_kept     = (val_true_means >= cut).mean()
    neg_removed = (val_neg_means  <  cut).mean()
    print(f"  cut={cut:3d} mm:  keeps {tp_kept:.1%} of true pairs,  removes {neg_removed:.1%} of neg pairs")

fig, ax = plt.subplots(figsize=(9, 5))
bins3 = np.linspace(0, 400, 70)
ax.hist(val_neg_means,  bins=bins3, alpha=0.45, color='steelblue', density=True,
        label=f'Negative pairs  (n={len(val_neg_means):,})')
ax.hist(val_true_means, bins=bins3, alpha=0.75, color='tomato',    density=True,
        label=f'True pairs  (n={len(val_true_means):,})')
ax.set_xlabel('Mean 2D position diff at overlapping layers (mm)', fontsize=13)
ax.set_ylabel('Density', fontsize=13)
ax.set_title(f'Per-radius position difference — full valset ({len(event_paths)} events)', fontsize=13)
ax.legend(fontsize=11)
fig.tight_layout(); plt.show()
```

```python

# --- Valset: Min 3D position diff at r > 150 mm (no fitting, all negatives) ---
# 
# "Same layer" is defined as |r_A - r_B| < R_TOL_MM = 15 mm — a proxy for same
# detector layer. Works as long as adjacent layers are spaced > 15 mm apart.
# A cleaner alternative would be to match by volume/layer ID if it's on the graph.
#
# Why r > 150 mm: at large r the outgoing and return arcs of a true pair have
# diverged angularly, so the 3D separation is largest there. Confusers are still
# co-located at all r. Using the MINIMUM across those layers is conservative:
# even the closest approach at high r should be non-zero for a true pair.

R_HIGH = 150.0   # mm — only consider hits above this radius

val3d_true_mins = []
val3d_neg_mins  = []

for path in tqdm(event_paths, desc='Valset min-3D-diff r>150'):
    graph = torch.load(path, map_location='cpu', weights_only=False)
    segs  = get_segments(graph)
    segs  = [s for s in segs if s.outer_r < OUTER_R_THRESHOLD]
    if len(segs) < 2:
        continue

    _hr = graph.hit_r.numpy()
    _hx = graph.hit_x.numpy()
    _hy = graph.hit_y.numpy()
    _hz = graph.hit_z.numpy()
    pids = np.array([get_segment_particle_id(s, graph) for s in segs])

    # Pre-sort by r, keep only hits with r > R_HIGH
    sorted_segs = []
    for s in segs:
        idx   = np.array(s.hits)
        order = np.argsort(_hr[idx])
        rs = _hr[idx][order]; xs = _hx[idx][order]
        ys = _hy[idx][order]; zs = _hz[idx][order]
        mask = rs > R_HIGH
        sorted_segs.append((rs[mask], xs[mask], ys[mask], zs[mask]))

    n_s = len(segs)
    for i in range(n_s):
        pid_i = int(pids[i])
        if pid_i == 0:
            continue
        ri, xi, yi, zi = sorted_segs[i]
        if len(ri) == 0:
            continue
        for j in range(i + 1, n_s):
            pid_j = int(pids[j])
            if pid_j == 0:
                continue
            rj, xj, yj, zj = sorted_segs[j]
            if len(rj) == 0:
                continue

            # Nearest-r match within R_TOL_MM, 3D distance
            dr = np.abs(ri[:, None] - rj[None, :])   # (Hi, Hj)
            nearest = np.argmin(dr, axis=1)
            mask = dr[np.arange(len(ri)), nearest] < R_TOL_MM
            if not mask.any():
                continue

            n_idx = nearest[mask]
            dx = xi[mask] - xj[n_idx]
            dy = yi[mask] - yj[n_idx]
            dz = zi[mask] - zj[n_idx]
            min_d = float(np.min(np.sqrt(dx**2 + dy**2 + dz**2)))

            if pid_i == pid_j:
                val3d_true_mins.append(min_d)
            else:
                val3d_neg_mins.append(min_d)

val3d_true_mins = np.array(val3d_true_mins)
val3d_neg_mins  = np.array(val3d_neg_mins)

print(f"Valset — true pairs with hit at r>{R_HIGH:.0f}mm:     {len(val3d_true_mins):,}")
print(f"Valset — negative pairs with hit at r>{R_HIGH:.0f}mm: {len(val3d_neg_mins):,}")
print()
for cut in [10, 20, 50, 100, 150]:
    tp_kept     = (val3d_true_mins >= cut).mean()
    neg_removed = (val3d_neg_mins  <  cut).mean()
    print(f"  cut={cut:4d} mm:  keeps {tp_kept:.1%} of true pairs,  removes {neg_removed:.1%} of neg pairs")

fig, ax = plt.subplots(figsize=(9, 5))
bins4 = np.linspace(0, 400, 70)
ax.hist(val3d_neg_mins,  bins=bins4, alpha=0.45, color='steelblue', density=True,
        label=f'Negative pairs  (n={len(val3d_neg_mins):,})')
ax.hist(val3d_true_mins, bins=bins4, alpha=0.75, color='tomato',    density=True,
        label=f'True pairs  (n={len(val3d_true_mins):,})')
ax.set_xlabel(f'Min 3D position diff at overlapping layers with r > {R_HIGH:.0f} mm (mm)', fontsize=12)
ax.set_ylabel('Density', fontsize=12)
ax.set_title(f'Min 3D diff at r>{R_HIGH:.0f}mm — full valset ({len(event_paths)} events)', fontsize=12)
ax.legend(fontsize=11)
fig.tight_layout(); plt.show()
```

## Part 3 — KDE Fit & Optimal Decision Boundary

The true-pair distribution is left-skewed with a hard pile-up near 1.0 — not Gaussian.
A **kernel density estimate (KDE)** is used instead to capture the actual shape of both
distributions, then the **Bayes-optimal decision boundary** is found where the weighted densities cross:

$$w_\text{true} \cdot \hat{p}(x\,|\,\text{true}) = w_\text{neg} \cdot \hat{p}(x\,|\,\text{neg})$$

Fitting is done on a 50 K subsample — KDE estimates are stable well below 4 M points.

## Part 4 — Hard Assignment with Optimal Boundary

Use the truncated-Gaussian optimal boundary as the cosine-similarity threshold for greedy segment matching, then evaluate efficiency and fake rate across all events.

> **Note**: This evaluates on `valset` (same data the fit was done on). Switch `DATA_DIR` to `testset` for an unbiased estimate.

```python

import sys
import pandas as pd
from pathlib import Path

_pipeline_root = Path().resolve().parent
if str(_pipeline_root) not in sys.path:
    sys.path.insert(0, str(_pipeline_root))

from low_pt_custom_utils.segment_matching import segments_to_track_labels
from low_pt_custom_utils.track_evaluation_utils import safe_evaluate_labelled_graph

# ── Config ────────────────────────────────────────────────────────────────────
BOUNDARY    = 0.70      # cosine-similarity threshold for greedy matching

USE_GEO_CUT = True      # toggle geometric pre-filter on/off
GEO_CUT_MM  = 200.0     # min 3D diff at r > R_HIGH must exceed this to be a candidate
# R_HIGH and R_TOL_MM defined in Part 1.5e cell above (150 mm, 15 mm)

node_scales = tuple(model.node_scales)

eval_config = dict(
    matching_fraction=0.7,
    matching_style='two_way',
    min_track_length=3,
    target_tracks={'pt': [0.1, float('inf')], 'nhits': [4, float('inf')]},
)

# ── Geometric pre-filter helper ───────────────────────────────────────────────
def passes_geo_cut(seg_i, seg_j, hr, hx, hy, hz,
                   r_high=R_HIGH, r_tol=R_TOL_MM, cut_mm=GEO_CUT_MM):
    """
    Returns True if the pair should be kept as a matching candidate.
    Computes the min 3D position diff at overlapping layers with r > r_high.
    If no overlapping high-r layers exist, the pair passes (no information → don't veto).
    """
    def _sorted(seg):
        idx   = np.array(seg.hits)
        order = np.argsort(hr[idx])
        rs = hr[idx][order]; mask = rs > r_high
        return rs[mask], hx[idx][order][mask], hy[idx][order][mask], hz[idx][order][mask]

    ri, xi, yi, zi = _sorted(seg_i)
    rj, xj, yj, zj = _sorted(seg_j)
    if len(ri) == 0 or len(rj) == 0:
        return True   # can't compute → don't veto

    dr      = np.abs(ri[:, None] - rj[None, :])
    nearest = np.argmin(dr, axis=1)
    mask    = dr[np.arange(len(ri)), nearest] < r_tol
    if not mask.any():
        return True   # no overlapping layers → don't veto

    n_idx  = nearest[mask]
    dx = xi[mask] - xj[n_idx]
    dy = yi[mask] - yj[n_idx]
    dz = zi[mask] - zj[n_idx]
    min_d = float(np.min(np.sqrt(dx**2 + dy**2 + dz**2)))
    return min_d >= cut_mm

# ── Evaluation loop ───────────────────────────────────────────────────────────
all_dfs = []
n_geo_vetoed = 0

for path in tqdm(event_paths, desc='Evaluating'):
    graph = torch.load(path, map_location='cpu', weights_only=False)

    segs     = get_segments(graph)
    complete = [s for s in segs if s.outer_r >= OUTER_R_THRESHOLD]
    to_match = [s for s in segs if s.outer_r < OUTER_R_THRESHOLD]

    matched_tracks, unmatched = [], complete + to_match

    if len(to_match) >= 2:
        seg_graphs = [segment_to_pyg(s, graph, node_scales=node_scales) for s in to_match]
        batch      = Batch.from_data_list(seg_graphs).to(device)
        with torch.no_grad():
            embs = model(batch.x, batch.edge_index, batch.batch)
        sim = (embs @ embs.T).cpu()

        hr = graph.hit_r.numpy()
        hx = graph.hit_x.numpy()
        hy = graph.hit_y.numpy()
        hz = graph.hit_z.numpy()

        n = len(to_match)
        pairs = []
        for i in range(n):
            for j in range(i + 1, n):
                s = sim[i, j].item()
                if s < BOUNDARY:
                    continue
                if USE_GEO_CUT and not passes_geo_cut(
                        to_match[i], to_match[j], hr, hx, hy, hz):
                    n_geo_vetoed += 1
                    continue
                pairs.append((s, i, j))
        pairs.sort(reverse=True)

        used = set()
        for s_val, i, j in pairs:
            if i not in used and j not in used:
                matched_tracks.append([to_match[i], to_match[j]])
                used |= {i, j}
        unmatched = complete + [to_match[i] for i in range(n) if i not in used]

    hit_t = graph.hit_t.cpu().numpy() if hasattr(graph, 'hit_t') else None
    graph.hit_track_labels = segments_to_track_labels(
        matched_tracks, unmatched, graph.hit_x.size(0), hit_t=hit_t
    )

    all_dfs.append(safe_evaluate_labelled_graph(
        graph,
        sel_conf=eval_config['target_tracks'],
        matching_fraction=eval_config['matching_fraction'],
        matching_style=eval_config['matching_style'],
        min_track_length=eval_config['min_track_length'],
    ))

eval_df = pd.concat(all_dfs, ignore_index=True)

parts = eval_df[eval_df['is_reconstructable']]
reco  = parts[parts['is_reconstructed'] & parts['is_matchable']]
trks  = eval_df[eval_df['is_matchable']]
mtchd = trks[trks['is_matched']]

n_p = len(parts.drop_duplicates(['event_id', 'particle_id']))
n_r = len(reco.drop_duplicates(['event_id', 'particle_id']))
n_t = len(trks.drop_duplicates(['event_id', 'track_id']))
n_m = len(mtchd.drop_duplicates(['event_id', 'track_id']))

eff_val   = n_r / n_p
frate_val = 1 - n_m / n_t

print(f"Boundary  = {BOUNDARY:.4f}  |  geo_cut = {'ON' if USE_GEO_CUT else 'OFF'}"
      + (f"  ({GEO_CUT_MM:.0f} mm, r>{R_HIGH:.0f}mm)" if USE_GEO_CUT else ""))
if USE_GEO_CUT:
    print(f"Geo-vetoed pairs: {n_geo_vetoed:,}")
print(f"  Efficiency:   {eff_val*100:.1f}%  ({n_r}/{n_p} particles)")
print(f"  Fake rate:    {frate_val*100:.2f}%  ({n_t - n_m}/{n_t} tracks)")
```

## Part 5 — Segments per Particle vs $p_T$

How many CC segments does each particle produce? Low-$p_T$ particles loop in the magnetic field → 2 segments (ingoing + outgoing arc). Higher-$p_T$ particles exit the detector entirely → 1 segment. The transition pT is where looping stops.

```python
from low_pt_custom_utils.track_evaluation_utils import PDG_TO_SPECIES

# ── Collect (pT, species, n_segs) for every reconstructable particle ──────────
seg_records = []

for path in tqdm(event_paths, desc='Counting segments per particle'):
    graph = torch.load(path, map_location='cpu', weights_only=False)

    # particle_id → pT and species
    pid_pt      = {}
    pid_species = {}
    if hasattr(graph, 'track_edges') and hasattr(graph, 'track_particle_pt'):
        edge_pids = graph.hit_particle_id[graph.track_edges[0, :]].numpy()
        for pid, pt_val in zip(edge_pids, graph.track_particle_pt.numpy()):
            if pid > 0 and pid not in pid_pt:
                pid_pt[int(pid)] = float(pt_val)
    if hasattr(graph, 'track_edges') and hasattr(graph, 'track_particle_type'):
        edge_pids = graph.hit_particle_id[graph.track_edges[0, :]].numpy()
        for pid, ptype in zip(edge_pids, graph.track_particle_type.numpy()):
            if pid > 0 and pid not in pid_species:
                pid_species[int(pid)] = PDG_TO_SPECIES.get(int(ptype), 'Other')

    # Count CC segments per particle (ALL segments, including complete tracks)
    pid_count = {}
    for seg in get_segments(graph):
        pid = get_segment_particle_id(seg, graph)
        if pid == 0:
            continue
        pid_count[int(pid)] = pid_count.get(int(pid), 0) + 1

    for pid, n_segs in pid_count.items():
        pt_val = pid_pt.get(pid)
        if pt_val is not None:
            seg_records.append({
                'pt':      pt_val,
                'species': pid_species.get(pid, 'Other'),
                'n_segs':  n_segs,
            })

segs_df = pd.DataFrame(seg_records)
print(f"Particles with segment data: {len(segs_df):,}")
print(segs_df['n_segs'].value_counts().sort_index().to_string())
```

## Part 7 — Proton CC Cluster Purity vs $p_T$

For every ground-truth proton that produces **exactly one CC segment**, compute the purity of that segment:

$$\text{purity} = \frac{\text{hits from the proton in the cluster}}{\text{total hits in the cluster}}$$

This happens before any segment matching — it shows whether the GNN is already contaminating proton clusters at the CC stage.

```python
proton_purity_records = []

for path in tqdm(event_paths, desc='Proton cluster purity'):
    graph = torch.load(path, map_location='cpu', weights_only=False)
    hit_pid = graph.hit_particle_id.numpy()

    # pid → pT and species
    pid_pt, pid_species = {}, {}
    if hasattr(graph, 'track_edges') and hasattr(graph, 'track_particle_pt'):
        edge_pids = graph.hit_particle_id[graph.track_edges[0, :]].numpy()
        for pid, pt_val in zip(edge_pids, graph.track_particle_pt.numpy()):
            if pid > 0 and pid not in pid_pt:
                pid_pt[int(pid)] = float(pt_val)
    if hasattr(graph, 'track_edges') and hasattr(graph, 'track_particle_type'):
        edge_pids = graph.hit_particle_id[graph.track_edges[0, :]].numpy()
        for pid, ptype in zip(edge_pids, graph.track_particle_type.numpy()):
            if pid > 0 and pid not in pid_species:
                pid_species[int(pid)] = PDG_TO_SPECIES.get(int(ptype), 'Other')

    # Build hit → segment index map
    segments = get_segments(graph)
    hit_to_seg = {}
    for seg_idx, seg in enumerate(segments):
        for h in seg.hits:
            hit_to_seg[h] = seg_idx

    # For each proton: find which segments contain its hits
    proton_pids = [pid for pid, sp in pid_species.items() if sp == 'Proton']
    for pid in proton_pids:
        pt_val = pid_pt.get(pid)
        if pt_val is None:
            continue

        # All hits belonging to this proton
        proton_hit_indices = np.where(hit_pid == pid)[0]
        if len(proton_hit_indices) == 0:
            continue

        # Which segments contain proton hits, and how many
        seg_hit_count = {}
        for h in proton_hit_indices:
            si = hit_to_seg.get(int(h))
            if si is not None:
                seg_hit_count[si] = seg_hit_count.get(si, 0) + 1

        if len(seg_hit_count) == 0:
            continue

        # Only single-segment protons (by hit presence, not majority vote)
        if len(seg_hit_count) != 1:
            continue

        seg_idx = next(iter(seg_hit_count))
        n_proton_in_seg = seg_hit_count[seg_idx]
        n_total_in_seg  = len(segments[seg_idx].hits)
        purity = n_proton_in_seg / n_total_in_seg

        proton_purity_records.append({
            'pt':      pt_val,
            'purity':  purity,
            'n_total': n_total_in_seg,
        })

purity_df = pd.DataFrame(proton_purity_records)
print(f"Single-segment protons: {len(purity_df):,}")
print(purity_df['purity'].describe())
```

```python
pt_bins = np.linspace(0.1, 0.35, 26)   # single-seg protons mostly below 0.3 GeV
xc = (pt_bins[1:] + pt_bins[:-1]) / 2
xe = (pt_bins[1:] - pt_bins[:-1]) / 2

mean_pur, err_pur, n_bin = [], [], []
for lo, hi in zip(pt_bins[:-1], pt_bins[1:]):
    b = purity_df[(purity_df['pt'] >= lo) & (purity_df['pt'] < hi)]['purity']
    if len(b) == 0:
        mean_pur.append(np.nan); err_pur.append(0); n_bin.append(0)
    else:
        mean_pur.append(b.mean())
        err_pur.append(b.std() / np.sqrt(len(b)))
        n_bin.append(len(b))

mean_pur = np.array(mean_pur)
err_pur  = np.array(err_pur)
n_bin    = np.array(n_bin)

fig, ax1 = plt.subplots(figsize=(9, 5))

proton_color = SPECIES_COLORS.get('Proton', 'red')
ax1.errorbar(xc, mean_pur, xerr=xe, yerr=err_pur,
             fmt=SPECIES_MARKERS.get('Proton', 'o'), color=proton_color,
             capsize=3, capthick=1.5, markersize=5, lw=1.8,
             label='Mean purity (proton hits / cluster hits)')
ax1.axhline(0.7, color='gray', linestyle='--', lw=1.5, label='Matching threshold (0.7)')
ax1.set_xlabel('$p_T$ [GeV]', fontsize=14)
ax1.set_ylabel('CC cluster purity', fontsize=14, color=proton_color)
ax1.tick_params(axis='y', labelcolor=proton_color)
ax1.set_ylim(0, 1.05)
ax1.set_xlim(pt_bins[0], pt_bins[-1])
ax1.legend(fontsize=11, loc='upper left')
ax1.grid(True, alpha=0.3)

ax2 = ax1.twinx()
ax2.bar(xc, n_bin, width=xe * 1.5, alpha=0.15, color='steelblue')
ax2.set_ylabel('Number of protons', fontsize=12, color='steelblue')
ax2.tick_params(axis='y', labelcolor='steelblue')
ax2.set_ylim(0)

ax1.set_title(
    'CC cluster purity for single-segment protons vs $p_T$\n'
    '(purity = proton hits / total hits in cluster)',
    fontsize=12,
)
fig.tight_layout()

out_path = PIPELINE_ROOT / 'data' / 'visuals' / 'proton_cluster_purity_vs_pt.pdf'
out_path.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(out_path, format='pdf', bbox_inches='tight')
print(f'Saved: {out_path}')
plt.show()
```

```python
proton_purity_records = []

for path in tqdm(event_paths, desc='Proton cluster purity'):
    graph = torch.load(path, map_location='cpu', weights_only=False)
    hit_pid = graph.hit_particle_id.numpy()

    # pid → pT and species
    pid_pt, pid_species = {}, {}
    if hasattr(graph, 'track_edges') and hasattr(graph, 'track_particle_pt'):
        edge_pids = graph.hit_particle_id[graph.track_edges[0, :]].numpy()
        for pid, pt_val in zip(edge_pids, graph.track_particle_pt.numpy()):
            if pid > 0 and pid not in pid_pt:
                pid_pt[int(pid)] = float(pt_val)
    if hasattr(graph, 'track_edges') and hasattr(graph, 'track_particle_type'):
        edge_pids = graph.hit_particle_id[graph.track_edges[0, :]].numpy()
        for pid, ptype in zip(edge_pids, graph.track_particle_type.numpy()):
            if pid > 0 and pid not in pid_species:
                pid_species[int(pid)] = PDG_TO_SPECIES.get(int(ptype), 'Other')

    # Build hit → segment index map
    segments = get_segments(graph)
    hit_to_seg = {}
    for seg_idx, seg in enumerate(segments):
        for h in seg.hits:
            hit_to_seg[h] = seg_idx

    proton_pids = [pid for pid, sp in pid_species.items() if sp == 'Proton']
    for pid in proton_pids:
        pt_val = pid_pt.get(pid)
        if pt_val is None:
            continue

        proton_hit_indices = np.where(hit_pid == pid)[0]
        n_proton_total = len(proton_hit_indices)
        if n_proton_total == 0:
            continue

        # Which segments contain proton hits?
        seg_hit_count = {}
        for h in proton_hit_indices:
            si = hit_to_seg.get(int(h))
            if si is not None:
                seg_hit_count[si] = seg_hit_count.get(si, 0) + 1

        n_segs_present = len(seg_hit_count)
        if n_segs_present == 0:
            continue  # all hits isolated (below score cut)

        # Best segment = the one containing the most proton hits
        best_seg_idx = max(seg_hit_count, key=seg_hit_count.get)
        n_proton_in_best  = seg_hit_count[best_seg_idx]
        n_total_in_best   = len(segments[best_seg_idx].hits)

        purity       = n_proton_in_best / n_total_in_best
        completeness = n_proton_in_best / n_proton_total

        proton_purity_records.append({
            'pt':           pt_val,
            'purity':       purity,
            'completeness': completeness,
            'n_segs':       n_segs_present,
            'n_total':      n_total_in_best,
        })

purity_df = pd.DataFrame(proton_purity_records)
print(f"Protons with ≥1 segment: {len(purity_df):,}")
print(purity_df[['purity', 'completeness']].describe())
```

## Secondaries

Identify secondary particles — those present in `hit_particle_id` but **not** in `track_edges` (outside the signal selection used when building ground-truth edges). In practice these are very short tracks: all secondaries have only **1 or 2 hits**.

A secondary is called **captured** if ≥ 50% of its hits belong to the same reconstructed track:
- 1-hit: captured if that hit lands in any track
- 2-hit: captured if at least 1 hit is in the same track

The cell re-runs the same track-building loop used in Part 4.

```python
# ── Diagnose particle structure in the first event ───────────────────────────
graph0 = torch.load(event_paths[0], map_location='cpu', weights_only=False)

all_pids    = torch.unique(graph0.hit_particle_id)
all_pids    = all_pids[all_pids > 0]

signal_pids = set()
if hasattr(graph0, 'track_edges'):
    signal_pids = set(graph0.hit_particle_id[graph0.track_edges.reshape(-1)].tolist())
    signal_pids.discard(0)

hit_pid = graph0.hit_particle_id

print(f"Total unique PIDs (excl noise) : {len(all_pids)}")
print(f"PIDs in track_edges (signal)   : {len(signal_pids)}")

# Show n_hits for ALL secondary particles across ALL events
sec_hit_counts = []
for path in event_paths:
    g = torch.load(path, map_location='cpu', weights_only=False)
    pids = torch.unique(g.hit_particle_id)
    pids = pids[pids > 0]
    sig  = set(g.hit_particle_id[g.track_edges.reshape(-1)].tolist()) if hasattr(g, 'track_edges') else set()
    sig.discard(0)
    for pid in pids.tolist():
        if pid not in sig:
            n = int((g.hit_particle_id == pid).sum().item())
            sec_hit_counts.append(n)

sec_hit_counts = sorted(sec_hit_counts)
print(f"\nSecondary particles across all {len(event_paths)} events: {len(sec_hit_counts)}")
if sec_hit_counts:
    from collections import Counter
    cnt = Counter(sec_hit_counts)
    print("  n_hits distribution:")
    for k in sorted(cnt):
        print(f"    {k} hits : {cnt[k]} particles")

del graph0
```

```python
MIN_SECONDARY_HITS = 1   # all secondaries have 1–2 hits
RECO_THRESHOLD     = 0.5  # ≥50% of hits in one track → "captured"
# For 1-hit: captured if the hit is in any track
# For 2-hit: captured if at least 1 hit is in the same track

secondary_records = []

for path in tqdm(event_paths, desc='Secondaries'):
    graph = torch.load(path, map_location='cpu', weights_only=False)

    # ── Rebuild tracks (same logic as Part 4) ────────────────────────────────
    segs     = get_segments(graph)
    complete = [s for s in segs if s.outer_r >= OUTER_R_THRESHOLD]
    to_match = [s for s in segs if s.outer_r < OUTER_R_THRESHOLD]

    matched_tracks, unmatched = [], complete + to_match

    if len(to_match) >= 2:
        seg_graphs = [segment_to_pyg(s, graph, node_scales=node_scales) for s in to_match]
        batch      = Batch.from_data_list(seg_graphs).to(device)
        with torch.no_grad():
            embs = model(batch.x, batch.edge_index, batch.batch)
        sim = (embs @ embs.T).cpu()

        n     = len(to_match)
        pairs = sorted(
            [(sim[i, j].item(), i, j)
             for i in range(n) for j in range(i + 1, n)
             if sim[i, j].item() >= BOUNDARY],
            reverse=True,
        )
        used = set()
        for s_val, i, j in pairs:
            if i not in used and j not in used:
                matched_tracks.append([to_match[i], to_match[j]])
                used |= {i, j}
        unmatched = complete + [to_match[i] for i in range(n) if i not in used]

    hit_t = graph.hit_t.cpu().numpy() if hasattr(graph, 'hit_t') else None
    graph.hit_track_labels = segments_to_track_labels(
        matched_tracks, unmatched, graph.hit_x.size(0), hit_t=hit_t
    )

    # ── Find secondary PIDs ───────────────────────────────────────────────────
    all_pids = torch.unique(graph.hit_particle_id)
    all_pids = all_pids[all_pids > 0]

    if hasattr(graph, 'track_edges'):
        signal_pids = set(
            graph.hit_particle_id[graph.track_edges.reshape(-1)].tolist()
        )
        signal_pids.discard(0)
    else:
        signal_pids = set()

    hit_pid   = graph.hit_particle_id
    hit_track = graph.hit_track_labels

    for pid in all_pids.tolist():
        if pid in signal_pids:
            continue

        mask   = hit_pid == pid
        n_hits = int(mask.sum().item())
        if n_hits < MIN_SECONDARY_HITS:
            continue

        assigned = hit_track[mask]
        assigned = assigned[assigned >= 0]

        if assigned.numel() == 0:
            secondary_records.append({'n_hits': n_hits, 'is_captured': False, 'purity': 0.0})
            continue

        _, counts  = torch.unique(assigned, return_counts=True)
        purity     = int(counts.max().item()) / n_hits

        secondary_records.append({
            'n_hits':     n_hits,
            'is_captured': purity >= RECO_THRESHOLD,
            'purity':     purity,
        })

sec_df = pd.DataFrame(secondary_records)
n_sec  = len(sec_df)
n_cap  = int(sec_df['is_captured'].sum())
print(f"Secondary particles (>= {MIN_SECONDARY_HITS} hit) across {len(event_paths)} events : {n_sec}")
print(f"  Captured (purity >= {RECO_THRESHOLD:.0%}) : {n_cap}  ({n_cap/max(n_sec,1):.1%})")
print(f"  Not captured                              : {n_sec - n_cap}  ({(n_sec-n_cap)/max(n_sec,1):.1%})")
print()
for nh, grp in sec_df.groupby('n_hits'):
    nc = int(grp['is_captured'].sum())
    print(f"  {nh}-hit secondaries: {len(grp)}  |  captured: {nc} ({nc/len(grp):.1%})")
```

```python
# ── Grouped bar: captured vs not-captured for 1-hit and 2-hit secondaries ────
hit_groups = sorted(sec_df['n_hits'].unique())
captured     = [int(sec_df[sec_df['n_hits'] == nh]['is_captured'].sum())     for nh in hit_groups]
not_captured = [int((~sec_df[sec_df['n_hits'] == nh]['is_captured']).sum())  for nh in hit_groups]
totals       = [c + nc for c, nc in zip(captured, not_captured)]

x    = range(len(hit_groups))
w    = 0.35
fig, ax = plt.subplots(figsize=(6, 4.5))

bars_c  = ax.bar([i - w/2 for i in x], captured,     width=w, color='steelblue', label='Captured')
bars_nc = ax.bar([i + w/2 for i in x], not_captured, width=w, color='salmon',    label='Not captured')

# Fraction labels on top of each bar
for i, (c, nc, tot) in enumerate(zip(captured, not_captured, totals)):
    ax.text(i - w/2, c + 1,  f'{c/tot:.0%}', ha='center', va='bottom', fontsize=11)
    ax.text(i + w/2, nc + 1, f'{nc/tot:.0%}', ha='center', va='bottom', fontsize=11)

ax.set_xticks(list(x))
ax.set_xticklabels([f'{nh}-hit secondaries\n(n={tot})' for nh, tot in zip(hit_groups, totals)], fontsize=12)
ax.set_ylabel('Count', fontsize=13)
ax.set_title(
    f'Secondary particle capture  (purity ≥ {RECO_THRESHOLD:.0%})\n'
    f'Overall: {n_cap}/{n_sec} = {n_cap/max(n_sec,1):.1%} captured',
    fontsize=12,
)
ax.legend(fontsize=11)
ax.grid(True, axis='y', alpha=0.3)
fig.tight_layout()
plt.show()
```
